```python
import json,random


with open('../positive_data.json','r',encoding='utf-8') as f:
    negative_words_data = json.load(f)

with open('../positive_data.json','r',encoding='utf-8') as f:
    positive_words_data = json.load(f)

filtered_negative_words = [word for word in negative_words_data]
filtered_positive_words = [word for word in positive_words_data]

random.seed(44)
sampled_negative_words = random.sample(filtered_negative_words, 100)
sampled_positive_words = random.sample(filtered_positive_words, 100)

with open('sampled_negative_words.json','w',encoding='utf-8') as f:
    json.dump(sampled_negative_words, f, ensure_ascii=False, indent=2)
with open('sampled_positive_words.json','w',encoding='utf-8') as f:
    json.dump(sampled_positive_words, f, ensure_ascii=False, indent=2)
```

```python
import openpyxl
from openpyxl.styles import Alignment, Font, Border, Side

# Combine positive and negative samples
all_words = sampled_positive_words + sampled_negative_words
# random.shuffle(all_words)

# Create a new workbook and select the active sheet
wb = openpyxl.Workbook()
ws = wb.active
ws.title = "Annotation"

# Define styles
alignment = Alignment(horizontal='center', vertical='center', wrap_text=True)
font = Font(name='等线', size=11)
border = Border(left=Side(style='thin'), 
                right=Side(style='thin'), 
                top=Side(style='thin'), 
                bottom=Side(style='thin'))

# Set column headers and styles
headers = ["标注", "词形", "构词结构", "释义", "例句", "语素义1", "语素义2"]
ws.append(headers)
for col_num, header in enumerate(headers, 1):
    cell = ws.cell(row=1, column=col_num)
    cell.alignment = alignment
    cell.font = font
    cell.border = border

# Set column widths
ws.column_dimensions['A'].width = 10
ws.column_dimensions['B'].width = 15
ws.column_dimensions['C'].width = 15
ws.column_dimensions['D'].width = 50
ws.column_dimensions['E'].width = 80
ws.column_dimensions['F'].width = 50
ws.column_dimensions['G'].width = 50
# Populate the sheet with data
current_row = 2
for word in all_words:
    start_row = current_row
    end_row = start_row + 4

    # Merge cells for the first four columns
    ws.merge_cells(start_row=start_row, start_column=1, end_row=end_row, end_column=1)
    ws.merge_cells(start_row=start_row, start_column=2, end_row=end_row, end_column=2)
    ws.merge_cells(start_row=start_row, start_column=3, end_row=end_row, end_column=3)
    ws.merge_cells(start_row=start_row, start_column=4, end_row=end_row, end_column=4)
    ws.merge_cells(start_row=start_row, start_column=6, end_row=end_row, end_column=6)
    ws.merge_cells(start_row=start_row, start_column=7, end_row=end_row, end_column=7)

    # Populate merged cells
    ws.cell(row=start_row, column=2, value=word['head_id'][0] + word['tail_id'][0])
    ws.cell(row=start_row, column=3, value=word['relation'])
    ws.cell(row=start_row, column=4, value=word['definition'])
    ws.cell(row=start_row, column=6, value=word['head_id'][0]+':'+word['head_sense'])
    ws.cell(row=start_row, column=7, value=word['tail_id'][0]+':'+word['tail_sense'])

    # Populate sentences and apply styles to all cells in the block
    sentences = word.get('sentences', [])
    for i in range(7):
        row_to_write = start_row + i
        # Write sentence if available
        if i < len(sentences):
            ws.cell(row=row_to_write, column=5, value=sentences[i])
        
        # Apply styles to all cells in the current row for this block
        for col in range(1, 8):
            cell = ws.cell(row=row_to_write, column=col)
            cell.font = font
            cell.border = border
            # Apply center alignment to all but the last column
            if col < 5:
                cell.alignment = alignment
            else: # Special alignment for sentence column
                cell.alignment = Alignment(wrap_text=True, vertical='center')

    current_row += 5

# Save the workbook
output_filename = 'annotation_data.xlsx'
wb.save(output_filename)

print(f"Annotation file '{output_filename}' created successfully with {len(all_words)} entries.")
```
