```python
import json,random
random.seed(44)
with open('../../data/3_MiCLS_processed.txt','r',encoding='utf-8') as f:
    lines = f.readlines()
    all_words = set([(line.split()[2],line.split()[4]) for line in lines])

with open('../../final_data_1124/positive_data.json', 'r', encoding='utf-8') as f:
    positive_data = json.load(f)
    positive_words = set((item['head_id'][0]+item['tail_id'][0],item['definition']) for item in positive_data)

with open('../../final_data_1124/negative_data_strict_filtered.json', 'r', encoding='utf-8') as f:
    negative_data = json.load(f)
    negative_words = set((item['head_id'][0]+item['tail_id'][0],item['definition']) for item in negative_data)
```

```python
import json
items={}
with open('../../data/3_MiCLS_processed.txt','r',encoding='utf-8') as f:
    lines = f.readlines()
    for line in lines:
        line=line.split()
        head_id = line[7]+line[6]
        tail_id = line[10]+line[9]
        definition = line[4] 
        relation = line[5]
        if (head_id+tail_id,definition) not in items:
            items[(head_id+tail_id,definition)] = {
                'head_id': head_id,
                'tail_id': tail_id,
                'definition': definition,
                'relation': relation
            }
    final_data = items.values()

with open('existing_word_with_definitions.json', 'w', encoding='utf-8') as f:
    json.dump(list(final_data), f, ensure_ascii=False, indent=4)
```

```python
existing_word_list = list(all_words)
existing_word_list = random.sample(existing_word_list, 1000)
positive_word_list = list(positive_words)
positive_word_list = random.sample(positive_word_list, 1000)
negative_word_list = list(negative_words)
negative_word_list = random.sample(negative_word_list, 1000)
```

```python
with open('word_list.json', 'w', encoding='utf-8') as f:
    json.dump({
        'existing_words': existing_word_list,
        'positive_words': positive_word_list,
        'negative_words': negative_word_list
    }, f, ensure_ascii=False, indent=4)
```
